# Social Media Impact on Students' Life

### 1. Introduction :-

This project analyzes how social media usage may be associated with students' sleep, stress, mental well-being, and academic performance.

### 2. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### 5. Data Cleaning

```python
df.isnull().sum()
```

```python
df["Perceived_Stress_Score"] = df["Perceived_Stress_Score"].fillna(
    df["Perceived_Stress_Score"].mean()
)

df["Academic_Performance_GPA"] = df["Academic_Performance_GPA"].fillna(
    df["Academic_Performance_GPA"].mean()
)
```

### 7. Data Visualization


##### Chart 1: Social Media Platform Usage

```python
plt.figure(figsize=(8,5))

sns.countplot(x="Primary_Platform", data=df)

plt.title("Social Media Platform Usage")
plt.xlabel("Platform")
plt.ylabel("Number of Students")
plt.show()
```

##### Chart 2: Social Media Usage vs Sleep Duration

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="Daily_Usage_Hours",
    y="Sleep_Duration_Hours",
    data=df
)

plt.title("Social Media Usage vs Sleep Duration")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Sleep Duration Hours")
plt.show()
```

##### Chart 3: GPA Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df["Academic_Performance_GPA"], bins=10, edgecolor="black")

plt.title("Distribution of Academic Performance GPA")
plt.xlabel("GPA")
plt.ylabel("Number of Students")
plt.show()
```

##### Chart 4: GPA Distribution by Overall Impact

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Overall_Impact",
    y="Academic_Performance_GPA",
    data=df
)

plt.title("GPA Distribution by Overall Social Media Impact")
plt.xlabel("Overall Impact")
plt.ylabel("Academic Performance GPA")
plt.show()
```

### 8. Key Findings

* The dataset contains 4,500 students and 16 columns.

* The average daily social media usage is approximately 5.30 hours per student.

* Instagram is the most commonly used platform, with 1,453 students, followed by TikTok with 1,226 students.

* Smartphones are the most commonly used devices, used by 3,775 students.

* The average sleep duration is approximately 6.70 hours per student.

* The average perceived stress score is approximately 13.51.

* The average GPA across different social media platforms is quite similar, ranging from approximately 3.42 to 3.43.

* Most students are classified under the Beneficial overall impact category (3,681 students), while 654 students are classified as Neutral and 165 as Negative.
